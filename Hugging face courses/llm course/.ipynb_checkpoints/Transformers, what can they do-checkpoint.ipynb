```python
from transformers import pipeline
sentimentAnalysisClassifier=pipeline("sentiment-analysis")
print(sentimentAnalysisClassifier("i am happy to say that This is a sample text i am writing to test the pipeline function in the transformers"))
```

```python
zeroShotClassifier=pipeline("zero-shot-classification")
print(zeroShotClassifier("i am happy to say that This is a sample text i am writing to test the pipeline function in the transformers",
                        candidate_labels=["education", "politics", "technology"]))
```

```python
textGenerator=pipeline("text-generation")
print(textGenerator("This is a good way"))
```

```python
textGenerator=pipeline("text-generation",model="distilgpt2")
print(textGenerator("excellence in dsa depends upon",max_new_tokens=30,num_return_sequences=2))
```

```python
#text generation using HuggingFaceTB/SmolLM2-360M
textGenerator=pipeline("text-generation",model="HuggingFaceTB/SmolLM2-360M")
print(textGenerator("In this course, we will teach you how to",
    max_length=30,
    num_return_sequences=2))
```

```python
unmasker=pipeline("fill-mask",model="bert-base-cased")
print(unmasker("i am [mask] who is studying in college",top_k=2))
```
