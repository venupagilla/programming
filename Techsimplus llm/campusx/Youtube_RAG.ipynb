```python
from youtube_transcript_api import YouTubeTranscriptApi
from langchain_classic.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_ollama import ChatOllama,OllamaEmbeddings
from langchain_core.prompts import PromptTemplate
```

```python
video_id=input("Enter the video id: ")
try:
    yt_transcripter=YouTubeTranscriptApi()
    transcript_list=yt_transcripter.fetch(video_id="kovF86_WqmQ")
    print("Transcript fetched successfully")
except Exception as e:
    print("Error fetching transcript:",e)
```

```python
text=" ".join([i.text for i in transcript_list])
print(text)
```

```python
text_splitter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=0)
text_chunks=text_splitter.create_documents([text])
print(len(text_chunks))
print(type(text_chunks))
print(text_chunks)
```

```python
llm=ChatOllama(model="llama3:8b")
embeddings=OllamaEmbeddings(model="llama3:8b")
vector_store=Chroma.from_documents(text_chunks,embedding=embeddings,persist_directory="C:\\Users\\pagil\\OneDrive\\Documents\\Programming\\Techsimplus llm\\campusx\\youtube_rag")
```

```python
query=input('Enter your question')
retrieved_docs=vector_store.similarity_search(query=query,k=2)
context=" ".join([i.page_content for i in retrieved_docs])
prompt_template=PromptTemplate.from_template(
    "you are  assistant who  answers the question from the context provided and if the context is empty or you cannot answer it based on the provided context just say i cant answer your question because the video doesnt contain information regarding the question asked question : {query} context : {context}"
)
prompt=prompt_template.invoke({"query": query,"context":context})
response=llm.invoke(prompt)
print(response.content)
```
