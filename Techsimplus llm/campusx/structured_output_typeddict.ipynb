```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from typing import TypedDict,Annotated,Literal
load_dotenv

class out1(TypedDict):
    title:Annotated[str,"This is the title of the para"]
    sentiment:Annotated[Literal["pos","neg"],"This is the sentiment of the para like negative,positve and neutral"]

model=ChatGroq(model="openai/gpt-oss-120b",temperature=1.0)

s_model=model.with_structured_output(out1)

```

```python
resp1 = s_model.invoke(
    "The 2006 Football League Championship play-off final was an association football match played on 21 May 2006 at the Millennium Stadium in Cardiff, Wales, between Leeds United and Watford. The match was to determine the third and final team to gain promotion from the Football League Championship, the second tier of English football, to the FA Premiership. The winners of two semi-finals competed for the final Premiership place of the 2006–07 season"
)
print(resp1)
```
