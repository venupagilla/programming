#### Text loader using langchain

```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader("../data/prabhas.txt",encoding="utf-8")
text1=loader.load()
print(text1[0].page_content)
```

### Loading data from webpage

```python
#we requrire beautifulsoup4 library to run this (WebBaseLoader uses beautifulsoup4 for working)
from langchain_community.document_loaders import WebBaseLoader
loader=WebBaseLoader(web_path="https://en.wikipedia.org/wiki/Prabhas")
web1=loader.load()
```

```python
print(len(web1))
```

```python
print(web1[0].page_content)
```

#### Wikipedia Loader in langchain

```python
from langchain_community.document_loaders import WikipediaLoader
loader=WikipediaLoader(query="prabhas" , load_max_docs=2)
wiki1=loader.load()
```

```python
print(len(wiki1))
```

```python
print(wiki1[0].page_content)
```
