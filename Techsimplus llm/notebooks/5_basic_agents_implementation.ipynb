```python
from dotenv import load_dotenv
from langchain_core.tools import tool
load_dotenv()
```

```python
@tool
def add_num(a:int,b:int)->int:
    """
    it will return the sum of 2 numbers

    Args:
        a:number one
        b:number two
    """
    return a+b

@tool
def mul_num(a:int,b:int)->int:
    """
    it will return the multiplication of 2 numbers

    Args:
        a:number one
        b:number two
    """
    return a*b
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
agent=create_agent(
    model=ChatGroq(model="llama-3.1-8b-instant"),
    tools=[add_num,mul_num],
    system_prompt="you are a math teacher, and always use tool for calcualtions"
)
```

```python
response=agent.invoke({"messages":[{
    "role":"user",
    "content":"what is 10*5+10*1000"
}]})
```
