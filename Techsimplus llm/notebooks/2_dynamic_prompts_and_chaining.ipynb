```python
#required imports
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import dotenv
dotenv.load_dotenv()
```

```python
#connecting google genai
model=ChatGoogleGenerativeAI(model="gemini-2.5-flash")
#creating output parset
outparser=StrOutputParser()
#testing the connection
model.invoke("HI")
```

```python
#creating a custom function to use inside the chain to convert the response into uppercase
def transform_val(x):
    return x.upper()
```

```python
#creating the dynamic prompt
prompts=ChatPromptTemplate.from_messages([
    {"role":"system","content":"You are a translator who translates the input to {language}"},
    {"role":"user","content":"{query}"}
])
print(prompts)
```

```python
final_prompt=prompts.format_messages(language="Hindi",query="I am a person who loves programming")
final_prompt
```

```python
model.invoke(final_prompt)
```

```python
#using chains to do the entire process above
chains=prompts | model | outparser | transform_val
res=chains.invoke({"language":"Hinglish","query":"python is best interms of versitality"})
print(res)
```
