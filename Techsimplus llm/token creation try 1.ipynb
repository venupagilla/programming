```python
import tiktoken
```

```python
text="Who invented zero?"
```

```python
tokenizer=tiktoken.encoding_for_model(model_name="gpt-4")
```

```python
tokenids=tokenizer.encode(text)
```

```python
print(tokenids)
```

```python
print(tokenizer.encode("Aryabhatta invented zero."))
```

```python
resultids=[32, 894, 370, 23976, 64, 36592, 7315, 13]
```

```python
result=tokenizer.decode(resultids)
```

```python
print(result)
```
